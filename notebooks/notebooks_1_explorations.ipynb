7 lazy evaluation : 
Spark n'exécute aucun calcul tant qu'une **action** : 
.show() , .collect() , .count() ou .write()


la différence entre une transformation et une action : 
Transformations (lazy)

- `filter()` : garde des lignes
- `withColumn()` : crée/modifie une colonne
- `select()` : choisit des colonnes

Actions

- `show()` : affiche les lignes
- `collect()` : rapatrie tout en Python
- `count()` : compte les lignes

8 : Spark UI

Spark UI (localhost:4041)

- **Job** : un travail déclenché par une action ,show(), count()
- **Stage** : une étape d'un job, découpée à chaque *shuffle* join, groupBy
- **Task** : la plus petite unité, traite une partition de données
- **Logs** : messages d'exécution (infos, warnings, erreurs) pour suivre et déboguer

1 action → 1 job → plusieurs stages → plusieurs tasks.

```python
# 9 : Spark vs Pandas
import pandas as pd
import time 

# Avec Pandas
t0 = time.time()
df_pandas = pd.read_csv("/home/jovyan/data/orders.csv")
print(f"Pandas : {time.time() - t0:.2f} s")

# Avec Spark
t1 = time.time()
df_spark = spark.read.csv("/home/jovyan/data/orders.csv", header=True)
print(f"Spark  : {time.time() - t1:.2f} s")
```

*Pandas est plus rapide car c'est un petit jeu de données.
